```python
text= ["Hi how are you", "iam fine", "thank you for asking"]
```

```python
text
```

```python
import re
re.split(r'\s',text[0])
```

```python
' '.join(re.split(r'\s',text[0]))
```

```python
allwords= re.split(r'\s',' '.join(text).lower())
allwords
```

```python
#find unique words
vocab = sorted(set(allwords))
vocab
```

```python
print(f'there are {len(allwords)} in the text and {len(vocab)} in the vocab ')
```

```python
#encoder and is python dictionary type
word2idx={}
for i,word in enumerate(vocab):
  word2idx[word]=i;
word2idx
```

```python
#decoder
idxtoword={}
for i,word in enumerate(vocab):
  idxtoword[i]=word
idxtoword
```

```python
print(f'the word "you" corresponds to {word2idx["you"]}')
print(f'the inex 7 corresponds to word "{idxtoword[7]}"')
```

```python
#select random word from the dictonary
import numpy as np
randidx=np.random.randint(0,len(vocab),size=5)
[idxtoword[i] for i in randidx]
```

```python
' '.join([idxtoword[i] for i in randidx])
```

```python
#translate text into numbers
text_as_int=[word2idx[word] for word in allwords]
text_as_int
```

```python
#and numbers back to text
for tokeni in text_as_int:
  print(f'Token {tokeni:2}: {idxtoword[tokeni]}')
```
